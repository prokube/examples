# Access MinIO with Python

Write the Iris dataset to MinIO and read it back.

```python
import pandas as pd
import s3fs
from sklearn.datasets import load_iris
```

```python
bucket = "<bucketname>"
key = "storage-examples/iris.csv"

if not bucket or bucket == "<bucketname>":
    raise ValueError("Set bucket to your MinIO bucket name.")
```

```python
df = load_iris(as_frame=True)
df.frame.head()
```

```python
s3 = s3fs.S3FileSystem()

with s3.open(f"{bucket}/{key}", "w") as f:
    df.frame.to_csv(f, index=False)

with s3.open(f"{bucket}/{key}", "rb") as f:
    read_back = pd.read_csv(f)

read_back.head()
```
